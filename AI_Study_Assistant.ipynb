```python
from google import genai
from google.colab import userdata

API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=API_KEY)

print("Gemini API connected successfully! ✅")
```

```python
response = client.models.generate_content(
    model="gemini-3.7-flash",
    contents="Say hello and explain Artificial Intelligence in one sentence."
)

print(response.text)
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Explain Artificial Intelligence in one simple sentence."
)

print(response.text)
```

```python
def ask_ai(prompt):
    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )
    return response.text
```

```python
print("========================================")
print("        📚 AI STUDY ASSISTANT")
print("========================================")

topic = input("\nEnter your topic or question: ")

print("\nChoose what you want:")
print("1. Explain Topic")
print("2. Summarize Topic")
print("3. Generate Quiz")

choice = input("\nEnter your choice (1/2/3): ")

if choice == "1":
    prompt = f"""
    Explain {topic} in simple language for a college student.
    Include:
    - Simple definition
    - Important points
    - One example
    """

elif choice == "2":
    prompt = f"""
    Give a short and easy-to-understand summary of {topic}.
    """

elif choice == "3":
    prompt = f"""
    Create 5 multiple-choice questions about {topic}.
    Give 4 options for each question and clearly show
    the correct answer.
    """

else:
    prompt = None
    print("❌ Invalid choice.")

if prompt:
    print("\n🤖 Generating answer...")

    answer = ask_ai(prompt)

    print("\n========================================")
    print("              🤖 AI RESPONSE")
    print("========================================\n")

    print(answer)
```

# 📚 AI Study Assistant

## About the Project

AI Study Assistant is an AI-powered mini project designed to help students learn and revise different topics using Google Gemini AI.

## Features

- 📖 Explain topics in simple language
- 📝 Summarize topics
- ❓ Generate multiple-choice quiz questions
- 🤖 AI-powered responses

## Technologies Used

- Python
- Google Gemini AI
- Google Colab

## Objective

The objective of this project is to understand how AI APIs can be integrated with Python to create a useful application for students.

## How It Works

1. Enter a topic or question.
2. Choose an option.
3. The request is sent to Gemini AI.
4. Gemini generates the response.
5. The response is displayed to the user.
